```python
#2. What is a Universal Function (UFunc) in NumPy? Provide an example.
#A Universal Function (UFunc) is a NumPy function that operates on ndarrays element-by-element.
import numpy as np
arr = np.array([1, 2, 3])
result = np.sqrt(arr)  # np.sqrt is a UFunc
print(result)
```

```python
#3. How do you create a 3x3 array of all zeros?
import numpy as np
zero_array = np.zeros((3, 3))
print(zero_array)
```

```python
#4. How do you create an array of 20 equally spaced points between 1 and 5?
import numpy as np
points_array = np.linspace(1, 5, 20)
print(points_array)
```
