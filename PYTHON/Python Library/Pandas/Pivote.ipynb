```python
import pandas as pd
```

```python
df = pd.read_csv(r"./sales_data.csv")
df
```

```python
df.pivot_table(index="Region", values="Sales_Amount",aggfunc="sum")
```

```python
df.pivot_table(index="Product_Category", values="Sales_Amount", columns="Region",aggfunc="sum")
```

```python
df.pivot_table(index="Sales_Rep", values="Quantity_Sold", columns=["Region","Sales_Channel"],aggfunc="sum")
```

```python
df.pivot_table(index="Sales_Rep", values="Sales_Amount", columns=["Region","Sales_Channel"],aggfunc=["count","sum","mean"])
```

```python
df.pivot_table(index="Payment_Method", values="Sales_Amount", columns="Customer_Type",aggfunc=["count","sum","mean","max"])
```

```python
df.pivot_table(index="Payment_Method", values="Unit_Price", columns="Region_and_Sales_Rep",aggfunc=["count","sum","mean","max"])
```

```python
df.pivot_table(index='Sales_Channel',values='Unit_Price',columns='Payment_Method')
```
