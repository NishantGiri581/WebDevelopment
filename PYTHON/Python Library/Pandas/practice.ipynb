```python
import pandas as pd
```

```python
df = pd.read_csv('./employee_productivity.csv')
df
```

```python
# 1. The HR manager wants to know which department has the highest average working hours. Use groupby to
# calculate the average Hours_Worked for each department.
df.groupby("Dept")["Hours_Worked"].mean()
```

```python
#2 The CEO wants a department-wise breakdown of total productivity. Calculate the total Tasks_Completed for
#. each department.

total_productivity= df.groupby("Dept")["Tasks_Completed"].sum().sort_values(ascending=False)
pd.DataFrame(total_productivity)
```

```python
# 3. The employee engagement team wants to identify the happiest department. Find the maximum
# Satisfaction_Rating per department.
df.groupby("Dept")["Satisfaction_Rating"].max()
```

```python
# 4. HR wants to know how many employees work in each department for manpower planning. Count employees
# per department.
df.groupby("Dept")["Emp_ID"].count()

```

```python
# 5. The operations head wants to review overall task distribution. Find the mean and sum of Tasks_Completed per
# department.
df.groupby("Dept")["Tasks_Completed"].agg(["mean","sum"])
```

```python
# 6. Management wants to evaluate which department is balanced in both efficiency and happiness. Compute
# average Hours_Worked and average Satisfaction_Rating per department.
df.groupby("Dept")[["Hours_Worked","Satisfaction_Rating"]].mean()
```

```python
# 7. HR wants to find underperforming units. Identify the minimum Hours_Worked in each department.
df.groupby("Dept")["Hours_Worked"].min()
```

```python
# 8. The finance team wants to understand work variability. Calculate the standard deviation of Hours_Worked for
# each department.
df.groupby("Dept")["Hours_Worked"].std()
```

```python
# 9. For award distribution, get the first employee (by order in dataset) from each department.
df.groupby("Dept").first()
```

```python
# 10. The productivity analyst wants the median number of tasks completed in each department.
df.groupby("Dept")["Tasks_Completed"].median()
```

```python
# 11. HR wants to know how many highly satisfied employees (Satisfaction_Rating ≥ 8) exist in each department.
df[df["Satisfaction_Rating"] >= 8].groupby("Dept").size()
```

```python
# 12. The performance manager wants to know how many employees in each department worked more than 8 hours.
df[df["Hours_Worked"] >= 8].groupby("Dept").size()
```

```python
# 13. The culture team wants to understand how satisfaction accumulates by department. Calculate total
# Satisfaction_Rating for each department.
df.groupby("Dept")["Satisfaction_Rating"].sum()
```

```python
# 14. HR wants to analyze satisfaction levels in detail. For each department, get the frequency of each
# Satisfaction_Rating.
df.groupby("Dept")["Satisfaction_Rating"].value_counts().unstack(fill_value=0)
```

```python
# 15. The CEO wants to meet the top performer in each department. Identify the employee with maximum
# Tasks_Completed per department using idxmax().
df.groupby("Dept")["Tasks_Completed"].max()
```

```python
# 16. Managers want to evaluate top contributors. List the top 3 employees with highest Hours_Worked per
# department.
df.sort_values("Hours_Worked", ascending=False).groupby("Dept").head(3)
```

```python
# 17. The analytics team wants to check if longer working hours actually increase productivity. Calculate correlation
# between Hours_Worked and Tasks_Completed within each department.
df.groupby("Dept").apply(lambda x:x["Hours_Worked"].corr(x["Tasks_Completed"]))
```
