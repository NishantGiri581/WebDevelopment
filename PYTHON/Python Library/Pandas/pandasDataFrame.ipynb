#MAP

```python
import pandas as pd
data = pd.read_csv("Titanic-Dataset.csv")
data
```

```python
gender = {"male" : "M", "female":"F"}
data['Sex']=data['Sex'].map(gender,)
```

```python
data
```

```python
embark = {"C":"Cherbourg", "Q":"Queenstown","S":"Southampton"}
data["Embarked"]=data["Embarked"].map(embark,)

```

```python
data
```

```python
#Apply Function => apply()
```

```python
data["Age"].isna().sum()
```

```python
data["Age"].fillna(round(data["Age"].mean(),0),inplace=True)
data
```

```python
data["Age"] = data["Age"].apply(lambda x:1 if x<1  else round(x,0))
```

```python
data["Age"].iloc[305]
```

```python

data["Age"] = data["Age"].astype("int")
```

```python
data
```
