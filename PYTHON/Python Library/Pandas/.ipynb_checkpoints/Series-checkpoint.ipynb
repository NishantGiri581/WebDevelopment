Solve The Problems Using Pandas Series
1. Stock Price Analysis
Analyze daily closing prices of a stock for a month.
Find:
Days when price dropped below average
Highest and lowest prices
% change from the first to the last day

```python
stock = {"01-11-2025":650,
        "02-11-2025":630,
        "03-11-2025":640,
        "04-11-2025":655,
        "05-11-2025":659,
        "06-11-2025":670,
        "07-11-2025":669,
        "08-11-2025":678,
        "09-11-2025":660,
         "10-11-2025":665,
         "11-11-2025":672,
         "12-11-2025":667,
        }

data = pd.Series(stock)
print(data)
```

```python
# Days when price dropped below average
price_dropped = data[data<data.mean()]
print(price_dropped.index)
```

```python
# Highest and lowest prices
Highest_Prices = data.max()
print(f"Highest_Prices : {Highest_Prices} ")
```

```python
Lowest_Prices = data.min()
print(f"Lowest_Prices : {Lowest_Prices} ")
```

```python
# % change from the first to the last day
x = data.head(1).values[0]
y = data.tail(1).values[0]
difference = y - x
percent = (difference/x)*100
percent
```

	
###  **2. E-commerce Daily Orders**
Given the number of orders received each day:
- Calculate weekly averages
- Detect sudden spikes or drops (greater than 30%)
- Find the most consistent week

```python
orders={"01-11-2025":50,
       "02-11-2025":30,
       "03-11-2025":40,
       "04-11-2025":55,
      "05-11-2025":59,
      "06-11-2025":70,
      "07-11-2025":69,
      "08-11-2025":78,
      "09-11-2025":60,
      "10-11-2025":65,
      "11-11-2025":72,
      "12-11-2025":67,}
order=pd.Series(orders)
order
```

```python
# - Calculate weekly averages
weekly_data = order.head(7)
avg_data = weekly_data.mean()
avg_data
```

```python
# - Detect sudden spikes or drops (greater than 30%)
change_rate = order.pct_change()*100
x = change_rate[change_rate > 30]
y = change_rate[change_rate < 30]
print("Order change_rate greater then 30% \n",x)
print("Order change_rate less then 30% \n",y)
```

3. Patient Body Temperature Monitoring

Monitor temperature data for a patient:

Detect fever patterns
Average daily temperature
Number of fever alerts (temperature > 100.4°F)


```python
import pandas as pd
```

```python
sales_data = [
    {"product":"Phone","quantity":20,"price":6000},
    {"product":"Laptop","quantity":5,"price":56000},
    {"product":"Keyboard","quantity":10,"price":500}
]
```

```python
df = pd.DataFrame(sales_data)
df
```

```python
df['total_price'] = df['quantity'] * df['price']
df['total_price'] 
```

```python
total_average = df['total_price'].mean()
```

```python
print(total_average)
```

```python
total_revenue = df['total_price'].sum()
```

```python
print(total_revenue)
```

```python
max_sell = df.loc[(df['quantity'] * df['price']).idxmax(), 'product']
```
