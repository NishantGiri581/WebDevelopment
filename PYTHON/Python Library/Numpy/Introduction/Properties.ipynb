```python
# .ndim:- It tells the number of dimensions of the array.

import numpy as np
arr_1d = np.array([1, 2, 3])
arr_2d = np.array([[1, 2], [3, 4]])
print(f"Dimensions of arr_1d: {arr_1d.ndim}") # Output: 1
print(f"Dimensions of arr_2d: {arr_2d.ndim}") # Output: 2
```

```python
# .shape:- It tells the shape of array and return tuple of integer indicating size.

import numpy as np
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(f"Shape of the matrix: {matrix.shape}") # Output: (2, 3)
```

```python
# .size:- It give the total number of elements.

import numpy as np
arr = np.array([[1, 2], [3, 4], [5, 6]])
print(f"Total number of elements: {arr.size}") # Output: 6
```

```python
# .dtype:-  it show the data type of elements of array.

import numpy as np
int_array = np.array([1, 2, 3])
float_array = np.array([1.0, 2.5, 3.7])
print(f"Data type of int_array: {int_array.dtype}")   # Output: int64
print(f"Data type of float_array: {float_array.dtype}") # Output: float64
```

```python
# .nbytes:- it returns total numbers of bytes consumed by array's data.

import numpy as np
arr = np.array([1, 2, 3, 4, 5], dtype=np.int64)
# An int64 is 8 bytes. 5 elements * 8 bytes/element = 40 bytes.
print(f"Memory size of the array (bytes): {arr.nbytes}") # Output: 40
```

```python
#2. Slicing and Views
"""
Create an array arr = np.array([10, 20, 30, 40]). 
Create a new variable view_arr by slicing arr to include elements 20 and 30. 
Check the .base property of view_arr.
"""
arr = np.array([10,20,30,40])
view_arr = arr[1:3]
print(f"Elements: {view_arr}")
print(f"Base of view_arr: {view_arr.base is arr}") # Output: True
```

```python
"""
Change the value of the first element in view_arr to 99. 
Print both arr and view_arr to show how the original array was affected.
"""
arr = np.array([10, 20, 30, 40])
view_arr = arr[1:3]
view_arr[0] = 99
print(f"Original array after change: {arr}")
print(f"View array after change: {view_arr}")
```

```python
"""
Create a copy of arr using .copy(). 
Store it in a new variable copy_arr. 
Check its .base property.
"""
arr = np.array([10, 20, 30, 40])
copy_arr = arr.copy()
print(f"Base of copy_arr: {copy_arr}")
```

```python
"""
Modify copy_arr by changing its last element to 100.
 Print arr and copy_arr to show they are independent.
"""
arr = np.array([10, 20, 30, 40])
copy_arr = arr.copy()
copy_arr[-1] = 100
print(copy_arr)
```

```python

"""
Use .nbytes and .itemsize to calculate the .size of an array.
 The formula is size = nbytes / itemsize. Write a script to prove this.
"""

arr = np.array([10, 20, 30], dtype=np.int32)

nbytes = arr.nbytes
itemsize = arr.itemsize

calculated_size = nbytes / itemsize

print(f"Size from properties: {arr.size}")
print(f"Size calculated from nbytes and itemsize: {calculated_size}")
```

```python
#Properties and Attributes of NumPy Arrays

# shape: The shape of the array represents the dimensions of the array. It returns a tuple indicating the number of elements in each dimension.

# ndarray: The core object of NumPy is the ndarray (N-dimensional array).

# dtype : The data type of the array elements.

# size : The total number of elements in the array.

# ndim : The number of dimensions (axes) of the array.

# itemsize : The size in bytes of each element of the array.

# nbytes : The total number of bytes consumed by the elements of the array.

# T : The transposed array (only for 2D arrays and higher).
```
