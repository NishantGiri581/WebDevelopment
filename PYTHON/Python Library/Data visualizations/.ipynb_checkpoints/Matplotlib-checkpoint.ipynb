```python
Min = df['total_bill'].min()
Max = df['total_bill'].max()
```

```python
Q1 = df['total_bill'].quantile(.25)
Q2 = df['total_bill'].quantile(.50)
Q3 = df['total_bill'].quantile(.75)
```

```python
IQR = Q3-Q1
```

```python
Lower_Limit = Q1 - 1.5*IQR
Upper_Limit = Q3 + 1.5*IQR
```

```python
print("Min :",Min)
print("Max :",Max)
print("Q1 :",Q1)
print("Q2 :",Q2)
print("Q3 :",Q3)
print("IQR :",IQR)
print("Lower_Limit :",Lower_Limit)
print("Upper_Limit :",Upper_Limit)
```

```python
df[df['total_bill']>Upper_Limit]
```
