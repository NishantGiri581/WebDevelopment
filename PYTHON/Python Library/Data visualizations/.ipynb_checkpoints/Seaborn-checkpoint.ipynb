```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import pyplot as plt
import seaborn as sns
import plotly_express as px
```

```python
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')
flights = sns.load_dataset('flights')
penguins = sns.load_dataset('penguins')
```

```python
#Histogram
sns.histplot(tips['total_bill'])
```

```python
#Histplot Kernal Density
sns.histplot(tips['total_bill'], kde=True,bins=30)
```

```python
#Kde Plot
sns.kdeplot(tips['total_bill'])
```

```python
#Density plot
sns.distplot(tips['total_bill'],hist=False)
```

```python
#Boxplot
sns.boxplot(x=tips['total_bill'],hue=tips['sex'])
```

```python
sns.boxplot(x='sex', y='total_bill', data=tips)
```

```python
#Violin plot
sns.violinplot(tips['tip'])
```

```python
sns.violinplot(tips['total_bill'])
```

```python
sns.scatterplot(x=tips['total_bill'], y=tips['tip']) #by varient
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=tips['total_bill'], y=tips['tip'], hue='sex', data=tips, size='size',style='time') #Multiple varients
```

```python
sns.lineplot(x='day',y='total_bill',hue='sex',data=tips)
```

```python
sns.relplot(x='day',y='total_bill',hue='sex',data=tips)
```

```python
sns.barplot(x='day',y='total_bill',data=tips)
```

```python
sns.countplot(x=tips['day'])
```

```python
sns.heatmap(tips.corr(numeric_only=True),annot = True)
```
