```python
#reverse of Sentence
def reverse_sent(s):
    word = s.split()
    rev_sent = " ".join(word[::-1])
    return rev_sent
print(reverse_sent("Python is Best")) 
```

```python
#Palindrome
def palindrome(s):
    s = s.lower()
    return s == s[::-1].lower() 
palindrome("Madam")    
    
```

```python
#Prime number
def is_prime(n):
    if n < 1:
        return False
    for i in range(2, int(n**0.5 ) + 1):  
        if n % i == 0:
            return False
    return True
print(is_prime(17)) 
```

```python
#Fibonnaci
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a , b = b, a+b
print(list(fibonacci(8)))         
```

```python
#Classmethhod 
class Student:
    school = "KCI"
    locations = "Hajipur"

    @classmethod
    def show_school(cls):
        print("School :", cls.school)
    @classmethod    
    def show_location(cls):
        print("Location :", cls.locations)
Student.show_school()
Student.show_location()
    
```

```python
#Instance method is used to specific object of class
class Student:
    def __init__(self,name):
        self.name = name
    def display(self):
        print("Student Name :", self.name)
s1 = Student("Nishant")
s1.display()
```

```python
#Static method does not use cls or self parameter
class Calculator:
    def add(a,b):
        return a+b
print(Calculator.add(5,6))        
```

```python
#clauser
def outer(x):
    def inner(y):
        return x+y
    return inner
add = outer(5)
print(add(3))
    
```
