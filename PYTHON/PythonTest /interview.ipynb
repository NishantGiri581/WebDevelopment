```python
#REVERSE OF STRING
def reverse_str(s):
    word = s.split()
    reverse_word = " ".join(word[::-1])
    return reverse_word
print(reverse_str("Nishant is Hero"))   
```

```python
#Palindrome 
def is_palindrome(s):
    s = s.lower()
    return s == s[::-1]
print(is_palindrome("Madam") )   
```

```python
#PRIME NUMBER
def is_prime(n):
    if n < 1:
        return False
    for i in range(2,int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True    
is_prime(17)         
  
```

```python
#FIBONACCI SERIES
def fibonnaci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
print(list(fibonnaci(5)) )       

        
```

```python
#MISSING NUMBER ARRAY
def missing_number(arr):
    n = len(arr)+1
    total_sum = n * (n+1) // 2
    actual_sum = sum(arr)
    return total_sum - actual_sum
print(missing_number([1,2,4,5]))    
    
```

```python
#REMOVE DUPLICATE
def rem_dup(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
print(rem_dup([1,3,5,4,3,6,8,3,0]))    
```

```python
#COMMON ELEMENT
def common_element(lst1 , lst2):
    return list(set(lst1) & set(lst2))
print(common_element([1,3,5,7,9],[1,2,3,4,5,6]))    
```

```python
#FACTORIAL
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))    
```

```python
#FIND DUPLICATE USING DICTIONARY
def find_duplicate(lst):
    freq = {}
    duplicate = []
    for num in lst:
        if num in freq:
            freq[num] += 1
            duplicate.append(num)
        else: 
            freq[num] = 1
    return duplicate
print(find_duplicate([1,2,3,2,5,2,7,1]))    
```

```python
#DECORATOR
def my_decorator(func):
    def wrapper(*args , **kwargs):
        print("Before Call")
        result = func(*args , **kwargs)
        print("After Call")
        return result
    return wrapper

@my_decorator
def greet(name):
    print("Hello",name)

greet("Nishant")
```

```python
import pickle

# 1. The Python object we want to save
student_data = {
    "name": "Nishant",
    "school": "KCI",
    "grades": [85, 90, 92]
}

# ==========================================
# PICKLING (Saving object to a file)
# ==========================================
# We must open the file in 'wb' mode (Write Binary)
with open("data.pkl", "wb") as file:
    pickle.dump(student_data, file)
    print("Data successfully pickled and saved!")

# ==========================================
# UNPICKLING (Loading object from a file)
# ==========================================
# We must open the file in 'rb' mode (Read Binary)
with open("data.pkl", "rb") as file:
    loaded_data = pickle.load(file)
    print("\nData successfully unpickled:")
    print(loaded_data)
    
    # It acts exactly like the original dictionary
    print("Student Name:", loaded_data["name"])
```

```python
#Pascal's Triangle
  #     1
  #    1 1
  #   1 2 1
  #  1 3 3 1
  # 1 4 6 4 1
n = 5
for i in range(n):
   print(" "* (n-i), end="")
   v = 1 
   for j in range(i + 1):
      print(v , end=" ")
      v = v * (i-j) // (j + 1)  
   print()
```

```python
# Butterfly Pattern
# *        *
# * *    * *
# * * *  * * *
# * * * ** * * *
# * * * ** * * *
# * * *  * * *
# * *    * *
# *        *

n = 5
for i in range(1, n + 1):
    print("*" * i, end="")
    print(" " * (2 * ( n - i )), end="")
    print("*" * i)

for i in range(n, 0, -1):
    print("*" * i, end="")
    print(" " * (2 * ( n - i )), end="")
    print("*" * i)
     
```

```python
#Hollow Square
# * * * * *
# *       *
# *       *
# *       *
# * * * * *

n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end =" ")
        else:
            print(" ",end=" ")
    print()        
```
