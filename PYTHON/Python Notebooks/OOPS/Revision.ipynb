```python
#Inheritance:- Child class inherit the behaviour of parent class
```

```python
class Animal :
    def __init__ (self,name,species):
        self.name = name
        self.species = species

    def make_sound(self):
        print("Generic Sound")   
```

```python
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(self, species = "Dog")
        self.breed = breed

    def make_sound(self):
         print("Woof!")
```

```python
class Cat(Animal):
    def __init__(self,name,color):
        super().__init__(self, species = "Cat")
        self.color = color

    def make_sound(self):
         print("Meow!")
```

```python
my_animal = Animal("Creature","Unknown") 
my_dog = Dog("Buddy","German Retriever") 
my_cat = Cat("Whiskers","Gray")
```

```python
my_animal.make_sound()
my_dog.make_sound() 
my_cat.make_sound()

print(f"{my_dog.name} is a {my_dog.species} of breed {my_dog.breed}.")
```

```python
#Abstraction:- It is a oop concept which hide the complex implimentation of code and show the necessary functionality to the user.
```

```python
from abc import ABC, abstractmethod
```

```python
class Remote(ABC):
    @abstractmethod
    def turn_on (self):
        pass
    def turn_off (self):
        pass    
class TvRemote(Remote):
    def turn_on (self):
        return "TV is ON" 
    def turn_off(self):
        return "TV is OFF"
remote = TvRemote()
print(remote.turn_on())  
print(remote.turn_off())  
```

```python
#Encapsulation:- It is a concept to grouping the data and method in a class and 
              # restrict direct access to the data, only authenticate one can access.
```

```python
class Person:
    def __init__(self,name,age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def set_name(self,name):
        self.__name = name

person = Person("Nishant",24)
print(person.get_name())
person.set_name("Muskan")
print(person.get_name())             
```

```python
#polymorphism:- poly means many, morph means shape, the ability of a program to execute differnet functionality based on context.
```

```python
class Circle:
    def draw(self):
        print("Drawing a circle: O")
        
class Square:
    def draw(self):
        print("Drawing a square: []")

class Triangle:
    def draw(self):
        rows = 5
        for i in range(1, rows + 1):
           print("*" * i)

# --- Let's draw some shapes ---

# Create a list of different shape objects
shapes = [Circle(), Square(), Triangle()]

# This loop doesn't care what type of object is in the list.
# It only cares that each object has a .draw() method. This is polymorphism!
for shape in shapes:
    shape.draw()
```
