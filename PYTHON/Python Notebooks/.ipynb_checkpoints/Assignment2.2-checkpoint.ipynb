4. Can you modify a tuple after its created ? Why or why not ?

*If you need to modify a sequence of elements, you should use a list instead, as lists are mutable. You can convert a tuple to a list, modify it, and then convert it back to a tuple if needed.*

```python
my_tuple = (1, 2, 3)
my_list = list(my_tuple)
my_list[0] = 4
my_tuple = tuple(my_list)
print(my_tuple)  # Output: (4, 2, 3)
```
