**String** :- It is a data typr that stores a sequence of character.

```python
# Concatenation
str1 = "Nishant"
str2 = "Giri"
str3 = str1 + " " + str2
print(str3)
#Length of string
len(str3)
```

```python
str = "Apna_college"
str[0] #indexing
```

```python
#indexing
s = "Python"
print(s[0])   # P
print(s[5])   # n
print(s[-1])  # n (negative index from end)
```

```python
#Slicing (get substring)
s = "Python"
print(s[0:3])
print(s[2:])
print(s[:4])
print(s[-3:])
```

```python
txt = "  hello Python  "

print(txt.upper())     # HELLO PYTHON
print(txt.lower())     # hello python
print(txt.title())     # Hello Python
print(txt.strip())     # removes spaces -> "hello Python"
print(txt.replace("Python", "World"))  # hello World
print(txt.split())     # ['hello', 'Python']
print("-".join(["Python", "is", "fun"]))  # Python-is-fun
```

**String Formating**

```python
name = "Nishant"
age = 21

print(f"My name is {name} and I am {age} years old.")
print("My name is {} and I am {} years old.".format(name, age))
```

**Usefull Operations**

```python
s = "Python Programming"

print(len(s))         # length
print("Python" in s)  # True
print("Java" not in s)  # True
print(s.count("m"))   # count occurrences
print(s.find("Prog")) # index of substring
```

**Joining & Reversing**

```python
words = ["Python", "is", "fun"]
print(" ".join(words))   # Python is fun

s = "Python"
print(s[::-1])   # nohtyP (reverse)
```

```python
# ✅ Exercise 1: Write a program to check if a string is a palindrome (same forward & backward).
s = "madam"
if s == s[::-1]:
  print("Palindrome")
else:
  print("Not Palindrome")
```

```python
# ✅ Exercise 2: Write a program that counts vowels in a given string.
s = "communication"
vowels = "aeiou"
count = 0
for ch in s:
    if ch in vowels:
        count += 1
print("Vowel count:",count)


# s = "Python Programming"
# vowels = "aeiouAEIOU"
# count = sum(1 for ch in s if ch in vowels)
# print("Vowel count:", count)
```

```python
# ✅ Exercise 3: Given a sentence, print the word with the maximum length.
sentence = "My name is Nishant Giri"
words = sentence.split()
longest_word = max(words, key=len)
print("Longest word:", longest_word)




# sentence = "Python programming is powerful"
# words = sentence.split()
# longest = max(words, key=len)
# print("Longest word:", longest)
```

```python
#21. check if two string are anagrams of each other.
str1 = "Hello"
str2 = "Python"

ana_grams = sorted(str1.lower()) == sorted(str2.lower())

if ana_grams:
    print(f"'{str1}' and '{str2}' are anagrams.")
else:
    print(f"'{str1}' and '{str2}' are NOT anagrams.")
```
