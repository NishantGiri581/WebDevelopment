```python
#Wap reverse a string in python.
def rev_string(s):
    return s[::-1]

rev_string("hello")    
```

```python
#Check palindrome in python.
def Palindrome(s):
    s = s.replace(" ","").lower()
    return s == s[::-1]

Palindrome("Madam") 
```
