```python
#Encapsulation  
#Encapsulation hides data from direct access.
class Account:
    def __init__(self, initial_balance=1000):
        self.__balance = initial_balance  # Private attribute

    # Getter: Controlled read access
    def get_balance(self):
        return self.__balance

    # Setter: Controlled write access with validation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ${amount}")
        else:
            print("Insufficient funds or invalid amount.")

# Usage
account = Account()
print("Current Balance:", account.get_balance())  # 1000
account.deposit(500)                              # Deposited: $500
account.withdraw(200)                             # Withdrew: $200
print("Updated Balance:", account.get_balance())  # 1300
```

```python
#Polymorphism  
# Same method name, different behavior.
class Bird:
    def sound(self):
        print("Bird Sound")

class Dog:
    def sound(self):
        print("Dog Sound")

bird = Bird()
dog = Dog()

bird.sound()
dog.sound()

```

```python
# Method Overriding  
# Child class can redefine a parent method.

# Example:
class Vehicle:
    def move(self):
        print("Vehicle Moving")

class Bike(Vehicle):
    def move(self):
        print("Bike Moving")

bike = Bike()
bike.move()

```

```python
import requests

# Example public API endpoint
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

print("Status Code:", response.status_code)  # e.g., 200

# Parse JSON response into a Python dictionary
data = response.json()
print("Received Data:", data)
print("Title:", data.get("title"))
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "product": "Laptop",
    "price": 50000,
    "available": True  # Capital 'T' in Python
}

response = requests.post(url, json=payload)

print("Status Code:", response.status_code)  # e.g., 201 (Created)
data = response.json()
print("Server Response:", data)
```
