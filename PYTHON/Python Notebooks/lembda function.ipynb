#Higher Order Function :- function which takes other function as argumenent.

```python
def list_operation(lst,fn):
  new = []
  for i in lst:
    new.append(fn(i))
  return new
```

```python
L1 = [1,2,3,4,5]
list_operation(L1,lambda x: x**2)
```

```python
float_value = lambda x:float(x)
list_operation(L1,float_value)
```

```python
#Change the list of string in a title case.
L = ["hello","world"]
title = lambda x: x.title()
list_operation(L,title)
```

#Tuple comprehension


```python
T=([4, 3, 5, 4, 6, 4, 4, 6, 8, 9])
#WAP to return tuple of square.
square = tuple(i**2 for i in T)
square
```

```python
#WAP to convert square of odd number.
T=([4, 3, 5, 4, 6, 4, 4, 6, 8, 9])
square = tuple(i**2 for i in T if i%2!=0)
square
```

```python
#Find the length of word using tuple comprehension.
T = "Nishant"
length = tuple(len(i) for i in T)
length
```

```python
#Get word starting with s.
sentence = "She sells seashells by the seashore and sings songs"
words_starting_with_s = tuple(word for word in sentence.split() if word.lower().startswith('s'))
print(words_starting_with_s)
```

```python
#reverse word in a list.
sentence = "She sells seashells by the seashore and sings songs"
words_starting_with_s = tuple(word[::-1] for word in sentence.split())
print(words_starting_with_s)
```

```python
#capatalize word ending with ing.
sentence = "She sells seashells by the seashore and sing songs and i am going to village"
words_starting_with_s = tuple(word.title() for word in sentence.split() if word.lower().endswith('ing'))
print(words_starting_with_s)
```

```python
#flatent a 2d list in 1d list.
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened_list = [item for row in matrix for item in row]
print(flattened_list)
```

```python
#filter palendrom in list.
palendrom = ["level", "hello", "radar", "world", "madam"]
palindromes = [word for word in palendrom if word == word[::-1]]
print(palindromes)
```

```python
#replace space with hyphen.
sentence = "Python is super simple"
words = tuple(word.replace(" ", "-") for word in sentence.split())
print(words)
```

```python
#Find comman element form two list.
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

common_elements = tuple(item for item in list1 if item in list2)
print(common_elements)
```

#Recursion :- Recursion is a programming technique where a function calls itself within its own definition to solve a problem. It works by breaking down a larger problem into smaller, self-similar subproblems.

```python
def func1(n):
  if n==0:
    return 0
  func1(n-1)
  print(n)
func1(10)
```

```python
def func2(n):
  if n==0:
    return 0
  print(n)
  func2(n-1)
func2(10)
```

```python
#Sum of the natural number up to n term.
def SUM(N):
  if N==0:
    return 0
  return N + SUM(N-1)
SUM(10)
```

```python
#Fibonacci series
def fibonacci(n):
  if n==0:
    return 0
  elif n==1:
    return 1
  else:
    return fibonacci(n-2)+fibonacci(n-1)
```

```python
import time
start_time = time.time()
start_time
fibonacci(40)
end_time = time.time()
difference = end_time-start_time
print(difference)
```

```python
for i in range(10):
  print(fibonacci(i))
```

```python
#Count the digit of number.
def count_digit(n):
  if n==0:
    return 0
  return 1 + count_digit(n//10)
count_digit(123)
```

```python
#Sum of digit of a number.
def sum_digit(n):
  if n==0:
    return 0
  return n%10 + sum_digit(n//10)
sum_digit(123)
```
